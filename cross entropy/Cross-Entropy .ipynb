# Cross-Entropy

- 모델이 예측한 확률분포가 실제 정답의 확률 분포와 얼마나 일치하는지를 수치화한 손실함수

| 완벽한 정답  | 실제[1,0] - 예측[1,0] | 분포가 완전히 일치하여 손실값은 0이 됨.  |
| --- | --- | --- |
| 완벽한 오답 | 실제[1,0] - 예측[0,1] | 분포가 극단적으로 어긋나기떄문에 손실값이 무한대로 치솟음. |
| 중간 오답 | 실제[1,0] - 예측[0.5,0.5] | 분포가 절반에 겹치므로 적당히 큰 손실값이 생김.  |
- 로그 함수 : 예측값이 1(정답)에 수렴하면 손실을 0으로 최소화하고, 0(오답)에 가까워질수록 손실을 무한대로 발산시킴.

### 회귀 (Regression) 의 손실함수

- MSE(평균 제곱 오차) : 정답과 예측값의 차이를 제곱하여 평균을 구함 , 예측값이 정답에 멀어질수록 오차가 제곱으로 커지므로 틀린정도를 직관적으로 반영함.
- MAE(평균 절대 오차) : 오차의 크기를 있는 그대로 평가하고 싶을떄 사용함. 오차(실제값 - 예측값)에 절대값을 씌워 음수를 제거한 뒤, 평균을 구함.

일반적으로 큰 오차를 빠르게 줄여야하는 경우에는 MSE를 사용하고, 데이터에 노이즈나 튀는값이 많은 모델은 MAE를 사용함. 

### 분류 (Classfication) 의 손실함수

- Hinge Loss(힌지 손실) : 모델이 정답을 단순히 맞히는것을 넘어, 정답과 오답 사이에 일정한 간격을 두고 확실하게 분류 했는지 측정함.
- Cross-Entropy(크로스 엔트로피) : 실제 정답 분포와 모델의 예측 분포 사이의 차이를 측정함. 오답에 대해 로그 스케일의 무한한 패널티를 부여하여 모델이 정답 확률을 최대한 높이도록 함.

### 크로스 엔트로피를 인위적으로 설계한 이유?

1. **경사 소실의 문제 해결**
- 시그모이드 함수는 입력값이 커지거나 작아지면 그래프가 평평해짐. 이떄 MSE를 손실함수로 쓰게되면 미분값이 0에 가까워져서 가중치 업데이트가 멈춰버림.
- 크로스 엔트로피에 로그 함수는 시그모이드의 평평한 부분을 수학적으로 상쇄시켜주기떄문에 오차가 클떄 기울기를 크게 유지해주어 폭발적 학습이 가능함.

1. **오답에 대한 강력한 패널티**
- MSE는 오차가 커지면서 거리의 제곱만큼만 벌금을 부여하지만, 크로스 엔트로피는 정답 확률이 0에 수렴하는 오답을 내면 무한대의 벌금을 부여하도록 설계

1. **최대 우도 측정(MLE)** 
- 모델이 내놓는 확률 분포가 실제 정답 분포와 일치할수록 우도가 높아짐.
- 각 확률을 곱하게 되면 값이 너무 작아지기 떄문에 계산이 불가능해짐 → log를 씌우면 곱셈이 덧셈으로 변하며 계산이 쉽도록 변환.
